```python
# String Slicing and Indexing

print("1. STRING CONCATENATION")

string1 = "Hello "
name = input("Enter your Name: ")

string3 = ", welcome to Python programming"

result = string1 + name + string3

print("\nComplete String:")
print(result)

print("2. STRINDG SLICING AND INDEXING")

# a. First character
print("\nFirst character:")
print(result[0])

# b. Last character
print("\nLast character:")
print(result[-1])

# c. First 5 characters
print("\nFirst 5 characters:")
print(result[:5])

# d. Last 11 characters
print("\nLast 11 characters:")
print(result[-11:])

# e. Reverse the string
print("\nReverse:")
print(result[::-1])

# f. Extract Python
print("\nWord Python:")
print(result[17:23])
```

```python
# String Methods
print("3.STRING METHODS")

strM = "Python beginner tutorial"

# a. Convert to uppercase
print("Uppercase:")
print(strM.upper())

# b. Convert to lowercase
print("\nLowercase:")
print(strM.lower())

# c. Capitalize
print("\nCapitalize:")
print(strM.capitalize())

# d. Count occurrences of character 't'
print("\nNumber of 't' characters:")
print(strM.count('t'))

# e. Replace Python with Machine Learning
print("\nAfter replacement:")
print(strM.replace("Python", "Machine Learning"))
```

```python
# TUPLES

print("4. TUPLES")

# Create tuples
t1 = (10, 20, 30)
t2 = (40, 50, 60)

# a. Concatenate
t_combine = t1 + t2

print("\na. Combined tuple:")
print(t_combine)

# b. Repeat 3 times
t_repeat = t_combine * 3

print("\nb. Tuple repeated 3 times:")
print(t_repeat)

# c. Third element
print("\nc. Third element:")
print(t_combine[2])

# d. First three elements
print("\nd. First three elements:")
print(t_combine[:3])

# e. Last three elements
print("\ne. Last three elements:")
print(t_combine[-3:])
```
